```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# Load your extracted CSV file
df = pd.read_csv('wine_data.csv')

# View the first few rows and column names
print("Data Overview:")
print(df.info())
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Simple Bar Chart with Matplotlib
plt.figure(figsize=(8, 5))
df['quality'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Wine Count by Quality Rating')
plt.xlabel('Quality')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
# Average alcohol content by wine quality
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='quality', y='alcohol', palette='viridis')
plt.title('Average Alcohol Content by Wine Quality')
plt.xlabel('Wine Quality')
plt.ylabel('Alcohol (%)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='alcohol', y='quality', orient='h', palette='magma')
plt.title('Alcohol Level by Quality')
plt.xlabel('Alcohol (%)')
plt.ylabel('Quality')
plt.show()
```

```python
# Get counts for a categorical column
counts = df['quality'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Wine Quality')
plt.show()
```

```python
counts = df['quality'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)

# Add center circle to turn it into a donut chart
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Wine Quality Proportions')
plt.show()
```

```python
# Grouped bar chart using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='quality', y='alcohol', hue='pH', ci=None)
plt.title('Alcohol Level by Quality with pH Breakdown')
plt.xlabel('Quality')
plt.ylabel('Alcohol (%)')
plt.legend(title='pH Level', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Correlation of all features with quality
correlations = df.corr()['quality'].sort_values(ascending=False)

print("--- Feature Correlation with Wine Quality ---")
print(correlations)
```

```python
# Calculate outliers using the Interquartile Range (IQR) method
def find_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)
    return series[(series < lower_bound) | (series > upper_bound)]

print("--- Outlier Counts Per Column ---")
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    outlier_count = len(find_outliers(df[col]))
    print(f"{col}: {outlier_count} outliers")
```
